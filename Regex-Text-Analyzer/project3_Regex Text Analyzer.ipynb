# Regex Text Analyzer

A simple Python project for finding different types of information in a text using Regular Expressions (Regex).

In this project, you can search for emails, phone numbers, URLs, dates, or even use your own custom regex.

```python
# find email using regular expression
def email(text):
    pattern = r'\w+(?:[-._]\w+)*@[a-zA-Z]+(?:\.[a-zA-Z]+)+'
    email= re.findall(pattern,text)
    return email
```

```python
# find phone_number in different common formats
def phone_number(text):
    pattern = r'\d{3,4}[\s.-]?\d{3}[\s.-]?\d{4}|\+\d{1,3}[\s-]?\d{3}[\s-]?\d{3}[\s-]?\d{4}|\(\d{3}\)\s\d{3,4}[-\s]\d{4}|00\d{1,3}[\s-]?\d{3}[\s-]?\d{3}[\s-]?\d{4}|\+\d{1,3}\s?\(\d{3}\)[\s-]?\d{3,4}[-\s]\d{4}'
    phone = re.findall(pattern,text)
    return phone
```

```python
# find url while excluding email addresses
def url(text):
    email_pattern = r'\w+(?:[-._]\w+)*@[a-zA-Z]+(?:\.[a-zA-Z]+)+'
    text = re.sub(email_pattern,'',text)
    
    pattern = r'(?<![\w@])(?:https?://|www\.)[^\s<>\[\],()]+|(?<![\w@])(?:[a-zA-Z0-9-]+\.)+[a-zA-Z]{2,}(?=[\s,]|$)' 
    url = re.findall(pattern,text)
    return url
```

```python
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December",
    "Jan","Feb","Mar","Apr","Jun","Jul","Aug","Sep","Sept","Oct","Dec"
]

month_pattern = '|'.join(months)

# find dates in different common formats
def date(text):
    pattern = rf'\d{{2,4}}[./-]\d{{2}}[./-]\d{{2,4}}|(?:{month_pattern})\s\d{{2}},\s\d{{4}}|\d{{2}}\s(?:{month_pattern})\s\d{{4}}'
    date = re.findall(pattern,text, re.IGNORECASE)
    return date
```

```python
# Mapping user choices to corresponding search function 
functions = {'email':email,'phone number':phone_number,'url':url,'date':date}


while True:

    # Ask user the type of info they want to find 
    user_choice = (input('What do you want to find in your text?\nEmail\nPhone number\nurl\ndate\nCustom regex\n')).lower()

    # allow the user to search using a custom regex
    if user_choice == 'custom regex':
        
        user_input = input('Enter your regex:')
        print(user_regex(user_text,user_input))
    
    # Run the function corresponding to the user's choice
    elif user_choice in functions:
        
        print(functions[user_choice](user_text))
        
    # Handle invalid choices    
    else:
        clear_output()
        print(f"Your input was {user_choice} and it's not in our options!\nPlease enter correct answer!!")
    # Ask whether the user wants to perform another search    
    if not ask_continue():
        break
```
