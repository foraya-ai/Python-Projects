# Blackjack Game:

This game is built using OOP

# Rules:

The game is played between a player and a computer dealer. The goal is to get as close to 21 as possible without going over.

The player can choose to Hit and receive another card or Stay to end their turn. After that, the dealer draws cards until beating the player or going over 21.

Number cards have their face value, face cards are worth 10, and an Ace can be worth 1 or 11.

```python
import random
# Card properties and values
suits = ('Hearts','Dimonds','Clubs','Spades')
ranks = ('Two','Three','Four','Five','Six','Seven','Eight','Nine','Ten','Jack','Queen','King','Ace')
values = {'Two':2,'Three':3,'Four':4,'Five':5,'Six':6,'Seven':7,'Eight':8,'Nine':9,'Ten':10,'Jack':10,'Queen':10,'King':10,'Ace':11}
```

```python
# this class represents a single playing card
class Card:

    def __init__(self,rank,suit):
        self.rank = rank.capitalize()
        self.suit = suit.capitalize()
        self.value = values[self.rank]
        
    def __str__(self):
        return f'{self.rank} of {self.suit}'
```

```python
# Testing card class
card=Card('three','hearts')
```

```python
# Creates deck of cards class:
class Deck:

    def __init__(self):
        
        self.all_cards = []
        # create all 52 cards in the dexk
        for suit in suits:
            for rank in ranks:
                # we are making an object:
                created_card =Card(rank,suit)
                self.all_cards.append(created_card)
    #shffling the deck 
    def shuffle(self):
        return random.shuffle(self.all_cards)
    #remove and return one card from the deck 
    def deal_one(self):
        return self.all_cards.pop()
```

```python
# player class and managing hand and bankroll
class Player():
    
    def __init__(self,name,bank_roll=100,bet=0):
        self.name = name
        self.bank_roll = bank_roll
        self.bet = bet
        self.hand = []
        
    # Add one or multiple cards to player's hand 
    def add_cards (self,cards):
        if type(cards) == type([]):
            self.hand.extend(cards)
            return self.hand
        else:
            self.hand.append(cards)
            return self.hand

    # check if player has enough money for the bet
    def bet_save_check(self,bet):

        if bet > self.bank_roll:
            print("You don't have enough money in the bank roll!")
        else:
            self.bet = bet
            print("Your bet was accepted let's start the game!")

    
    def hand_value (self):
        # calculate the total value of player's hand
        myvalue=0
        ace_count=0
        for card in self.hand:    
            myvalue +=card.value

            if card.rank=='Ace':
                ace_count+=1
        # change Ace value from 11 to 1 if the hands exceeds 21        
        while myvalue>21 and ace_count>0:
            myvalue-=10
            ace_count-=1
        return myvalue
            
    def clear_hand(self):
        # clear all cards from player's hand
        self.hand.clear()
        return self.hand
```

```python
# Dealer class that inherits common attributes and methods from player
class Dealer(Player):
    def __init__(self):
        super().__init__("dealer")
```

```python
# Testing dealer
dealer=Dealer()
```

```python
# Main game logic:
print("Welcome to blackjack game!")

# create and shuffle new_deck
new_deck = Deck()
new_deck.shuffle()

# create player and dealer
player = Player('Jose')
dealer = Dealer()

# Get player's bet
player_bet = int(input('How much money do you want to bet in this game?'))
player.bet_save_check(player_bet)

# Deal two cards to both player and dealer
for x in range(2):
    player.add_cards(new_deck.deal_one())
    dealer.add_cards(new_deck.deal_one())

# display the initial hands 
print(f'{player.name} :\n{player.hand[0]}\n{player.hand[1]}\n')
print(f'{dealer.name} :\n{dealer.hand[0]}\n{dealer.hand[1]} ')

# get player first decision
player_choice = input(f'{player.name} total value is {player.hand_value()}.\nDo you want to hit or stay?').upper()

# player's turn
while player_choice == 'HIT':

    # Deal a new card to the player
    player.add_cards(new_deck.deal_one())
    print(player.hand[-1])

    # Check if the player busts
    if player.hand_value()>21:
        player.bank_roll-=player.bet
        print(f'The dealer has won!{player.hand_value()} your bank_roll is {player.bank_roll}')
        break
        
    # Ask the player to hit or stay again    
    elif player.hand_value()<21:
        player_choice = input(f'{player.name} total vaue is {player.hand_value()}!\nDo you want to hit or stay?').upper()

# Dealer's turn   
while player_choice == 'STAY':

    print('Dealer turn!')
    dealer_total=dealer.hand_value()

    # Dealer wins if their hand beats the player without busting
    if dealer_total<=21 and dealer_total>=player.hand_value():
        print("The dealer has won the game")
        player.bank_roll-=player.bet
        print(player.bank_roll)
        break 
        
    # Player wins if the dealer busts
    elif dealer_total > 21:
        print("Busts!The player has won the game")
        player.bank_roll+=player.bet
        print(player.bank_roll)
        break  
        
    # Dealer takes another card
    else:
        dealer.add_cards(new_deck.deal_one())
        print(dealer.hand[-1])
```
