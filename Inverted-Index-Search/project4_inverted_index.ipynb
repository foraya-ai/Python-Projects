# Text Search and Word Counter

A simple Python project for searching and counting words in multiple text files.

The text is cleaned by removing punctuation and stop words, then the user can search for a word and see which files contain it and how many times it appears.

```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import os
```

```python
# files paths:
repeated_path = "C:\\Users\\ALFA\\Documents\\"

data_science_path = "data/data_science.txt"
python_path = "data/python.txt"
software_path = "data/software_engineering.txt"
ai_path = "data/ai.txt"
web_path = "data/web.txt"

#store all files paths in a list 
file_list = [data_science_path,python_path,software_path,ai_path,web_path]    
```

```python
# read files and put them in dictionary
files_content = {}

for path in file_list:
    with open(path,'r',encoding='utf-8')as f:
        # file name extraction
        files_content[os.path.basename(path)] = f.read()
    
# print(files_content)
```

```python
# list of file names:
file_names = []

for path in file_list:
    file_names.append(os.path.basename(path))

print(file_names)
```

```python
# Prepare punctuation and English stop words for text cleaning

translator = str.maketrans('','',string.punctuation)
stop_words = set(stopwords.words('english'))

tokens_list = []
tokens_dic_re = {}

# cleaning and tokenizing text from each file 
for f_name,text in files_content.items():
    
    # deleting punctuations : 
    clean_text = text.translate(translator)
    
    # tokenizing text:
    tokens = word_tokenize(clean_text.lower())
    
    # deleting stop words:
    filtered_tokens = [word for word in tokens if word not in stop_words]
    tokens_list.append(filtered_tokens)
    
# print(tokens_list)
    
```

```python
tokens_dic_re = {}
# dictionary for cleaned text:
for file_name, file_token in zip(file_names,tokens_list):
    tokens_dic_re.update({file_name: file_token})
# print(tokens_dic_re)
```

```python
# creating a list for unique words for each file 
unique_tokens = []

for tokens in tokens_list:
    unique_tokens.append(list(set(tokens)))
# print(unique_tokens)

# store unique words for each file
unique_tokens_dic = {}

for file_name, file_token in zip(file_names,unique_tokens):
    unique_tokens_dic.update({file_name: file_token})

# print(unique_tokens_dic)
```

```python
# Get the word the user wants to search for
while True:
    
    user = input('enter the word you want to find?\n').lower()
    
    # count how many times the selected word appears in each file 
    def counting (word):
        for key,value in tokens_dic_re.items():
            count = value.count(word)
            print(f"file {key} has :{count} {word} words\n")
            
    # Check which files contain the searched word    
    for key,value in unique_tokens_dic.items():
        for word in value:
            if word == user:
                print(f"The **{word}**word is in file:{key}\n")
    print('---------------------------------------------------------')
    
    # Count the searched word in all files
    counting(user)
    
    while True:
        play_again = input('Do you want to continue searching? yes or no').lower() 
        if play_again == 'yes':
            break
        elif play_again == 'no':
            break
        else:
            print('Please enter yes or no.')

    # Exit the search loop if the user chooses no
    if play_again == 'no':
        break
```
